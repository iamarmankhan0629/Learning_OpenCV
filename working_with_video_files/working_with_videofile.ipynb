```python
import cv2
```

```python
# load the video
video = cv2.VideoCapture("./mountain.mp4")
```

```python
# display the video
while True:
    ret, frame = video.read()
    if not ret:
        break
    cv2.imshow("Video frame", frame)
    if cv2.waitKey(1) & 0xff == ord('q'):      # to break out from while loop when q is press
        break

# t0 ---> f1
# wait for 1 ms (waitkey(1))
# t1 ---> f2    

video.release()
cv2.destroyAllWindows()    
```

```python
# write a video
video = cv2.VideoCapture("./mountain.mp4")

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
hieght = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter("./output.mp4",fourcc,30,(width,hieght))

while True:
    ret, frame = video.read()
    if not ret:
        break
    out.write(frame)
    
out.release()
video.release()
cv2.destroyAllWindows()   
    
```

```python
# write a video
video = cv2.VideoCapture("./mountain.mp4")

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
hieght = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter("./output_new.mp4",fourcc,30,(width,hieght))

while True:
    ret, frame = video.read()
    if not ret:
        break
    frame = frame + 20
    out.write(frame)
    
out.release()
video.release()
cv2.destroyAllWindows()   
    
```
