```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("../Images/giraffe-Kenya.png")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_grey = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
hist, bins = np.histogram(image_grey.flatten(),256,[0,256])

# perform histogram equalization

equalized_image = cv2.equalizeHist(image_grey)

equalized_hist, bins = np.histogram(equalized_image.flatten(),256,[0,256]) 

# plot the results

plt.figure(figsize=(20,16))

plt.subplot(2,2,1)
plt.imshow(image_grey, cmap="grey")
plt.title("orginal image")


plt.subplot(2,2,2)
plt.plot(hist, color="blue")
plt.title("orginal image Histogram")


plt.subplot(2,2,3)
plt.imshow(equalized_image, cmap="grey")
plt.title("Equalized image")


plt.subplot(2,2,4)
plt.plot(equalized_hist, color="red")
plt.title("Equalized image Histogram")


plt.tight_layout()
plt.show()
 
```

```python
image_yuv = cv2.cvtColor(image, cv2.COLOR_BGR2YUV)

image_yuv[:,:,0] = cv2.equalizeHist(image_yuv[:,:,0])

equalized_color_image = cv2.cvtColor(image_yuv, cv2.COLOR_YUV2BGR)

equalized_color_image_rgb = cv2.cvtColor(equalized_color_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(20,16))

plt.subplot(1,2,1)
plt.imshow(image_rgb)
plt.title("orginal image")


plt.subplot(1,2,2)
plt.imshow(equalized_color_image_rgb)
plt.title("equalized image rgb")




plt.tight_layout()
plt.show()
```
