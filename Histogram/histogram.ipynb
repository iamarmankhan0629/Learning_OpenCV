```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("../Images/giraffe-Kenya.png")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)

```

```python
channels = cv2.split(image_rgb)
colors = ["red", "green", "blue"]

for channel, color in zip(channels,colors):
    histogram = cv2.calcHist([channel],[0],None,[256],[0,256])
    plt.plot(histogram,color= color)

plt.title("Color Histogram")
plt.xlabel("pixel intensity")
plt.ylabel("frequence")    
plt.show()    
```

## HSV

```python
from IPython.core.pylabtools import figsize
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

h, s, v = cv2.split(image_hsv)

h_histogram = cv2.calcHist([h],[0],None,[180],[0,180])

s_histogram = cv2.calcHist([s],[0],None,[256],[0,256])

v_histogram = cv2.calcHist([v],[0],None,[256],[0,256])

fig, axes = plt.subplots(1,3,figsize=(15,5))

axes[0].plot(h_histogram, color='orange')
axes[0].set_title("Hue Histogram")
axes[0].set_xlabel("Bins")
axes[0].set_ylabel("frequency")    
    
axes[1].plot(s_histogram, color='green')
axes[1].set_title("Saturation Histogram")
axes[1].set_xlabel("Bins")
axes[1].set_ylabel("frequency")  

axes[2].plot(v_histogram, color='blue')
axes[2].set_title("value Histogram")
axes[2].set_xlabel("Bins")
axes[2].set_ylabel("frequency")      

plt.tight_layout()
plt.show()
```
