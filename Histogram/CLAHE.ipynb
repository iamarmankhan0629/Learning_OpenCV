```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python

image = cv2.imread("../Images/giraffe-Kenya.png") # replace the image with sample.png

image_grey = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
equalized_image_grey = cv2.equalizeHist(image_grey)

clahe = cv2.createCLAHE(clipLimit=2.0,tileGridSize=(4,4))

clahe_image = clahe.apply(image_grey)

# plotting

plt.figure(figsize=(10,8))

plt.subplot(1,3,1)
plt.imshow(image_grey, cmap="gray")
plt.title("original image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(equalized_image_grey, cmap="gray")
plt.title("equalized image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
image_yuv = cv2.cvtColor(image, cv2.COLOR_BGR2YUV)

image_yuv[:,:,0] = clahe.apply(image_yuv[:,:,0])

yuv_image_bgr = cv2.cvtColor(image_yuv, cv2.COLOR_YUV2BGR)

yuv_image_rgb = cv2.cvtColor(yuv_image_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,8))

plt.subplot(1,2,1)
plt.imshow(image_rgb)
plt.title("original image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(yuv_image_rgb)
plt.title("equalized image")
plt.axis("off")


plt.tight_layout()
plt.show()

```
