```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("../Images/giraffe-Kenya.png")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

## flipping

```python
# 1 --> horizontally
# 0 --> vertically
# -1 --> both axes
flip_horizontal = cv2.flip(image, 1)
flip_vertical = cv2.flip(image, 0)
flip_bothaxes = cv2.flip(image, -1)

titles = ["original image", "horizontal flip", "vertical flip", "flip both axes"]
images = [image_rgb, 
          cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB), 
          cv2.cvtColor(flip_vertical, cv2.COLOR_BGR2RGB),
          cv2.cvtColor(flip_bothaxes, cv2.COLOR_BGR2RGB)]

plt.figure(figsize=(12,8))

for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')
    
    
plt.tight_layout()
plt.show()    
    
```

## rotating

```python
rotate_90_clockwise = cv2.rotate(image_rgb, cv2.ROTATE_90_CLOCKWISE)
rotate_90_counterclockwise = cv2.rotate(image_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)
rotate_180 = cv2.rotate(image_rgb, cv2.ROTATE_180)

titles = ["original image", "90 clockwise", "90 counter clockwise","180 degree"]
images = [image_rgb, rotate_90_clockwise, rotate_90_counterclockwise, rotate_180]

plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()    
```

## Cropping

```python
print(image_rgb.shape)
```

```python
x_start, y_star = 400, 300
x_end, y_end = 800, 900

cropped_image = image_rgb[y_star:y_end,x_start:x_end]

plt.figure(figsize=(8,8))

plt.subplot(2,2,1)
plt.imshow(image_rgb)
plt.title("original image")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(cropped_image)
plt.title("cropped Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```
