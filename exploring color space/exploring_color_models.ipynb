```python
import cv2
```

```python
bgr_image = cv2.imread("./mountain.jpg")
```

```python
# convert to grayscale
grayscale_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2GRAY)
cv2.imwrite("./gray_image.jpg",grayscale_image)
```

```python
# convert to HSV
hsv_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2HSV)
cv2.imwrite("./hsv_image.jpg",hsv_image)
```

```python
# convert to LAB
lab_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2LAB)
cv2.imwrite("./lab_image.jpg",lab_image)
```

```python
# convert to rgb
rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)
cv2.imwrite("./rgb_image.jpg",rgb_image) # this will give a bgr image as output
```

```python
cv2.imwrite("./bgr_image.jpg", bgr_image) # this wil give rgb as output


# when we give a rgb to output from cv it will give us bgr
# and when we give a bgr to output from cv it will give us rgb
```
