```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("../Images/giraffe-Kenya.png")

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
kernel_s = 15

average_blur = cv2.blur(image_gray,(kernel_s,kernel_s))

plt.imshow(average_blur, cmap="grey")
```

```python
kernel_s = 15

gaussian_blur = cv2.GaussianBlur(image_gray,(kernel_s,kernel_s),0)

plt.imshow(gaussian_blur,cmap="grey")
```

```python
median_blur = cv2.medianBlur(image_gray,11)

plt.imshow(median_blur,cmap="grey")
```
