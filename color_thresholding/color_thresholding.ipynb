```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("../images/giraffe-Kenya.png")
```

```python
rgb_image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
hsv_image = cv2.cvtColor(image,cv2.COLOR_BGR2HSV)
```

```python
plt.imshow(rgb_image)
```

```python
# setting up target . Let say we want to target the white pattern on giraffe
lower_white = np.array([0,0,190])
upper_white = np.array([30, 80, 255]) # because the pattern is not pure white at every point
```

```python
mask = cv2.inRange(hsv_image, lower_white, upper_white)
```

```python
plt.imshow(mask)
```

```python
black_hsv = np.uint8([[[120, 255, 0]]])
black_bgr = cv2.cvtColor(black_hsv, cv2.COLOR_HSV2BGR)[0][0]
```

```python
image_result = image.copy()
image_result[mask>0] = black_bgr
```

```python
image_result_rgb = cv2.cvtColor(image_result,cv2.COLOR_BGR2RGB)
```

```python
# plot 
 
fig,axs = plt.subplots(1, 3, figsize =(12, 6)) 

axs[0].imshow(rgb_image)
axs[0].set_title("original_image")
axs[0].axis('off')

axs[1].imshow(mask)
axs[1].set_title("binary mask")
axs[1].axis('off')

axs[2].imshow(image_result_rgb)
axs[2].set_title("output_image")
axs[2].axis('off')

plt.tight_layout()
plt.show()
 
```
