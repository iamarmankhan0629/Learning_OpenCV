```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("../Images/giraffe-Kenya.png", cv2.IMREAD_GRAYSCALE)

# Sobel edge detection

sobel_x = cv2.Sobel(image,cv2.CV_64F,1,0,ksize=3)
sobel_y = cv2.Sobel(image,cv2.CV_64F,0,1,ksize=3)

sobel_combine = cv2.magnitude(sobel_x,sobel_y)

# Laplacian edge detection

laplacian = cv2.Laplacian(image,cv2.CV_64F,ksize=3)
laplacian_ab = cv2.convertScaleAbs(laplacian)


#Canny

canny = cv2.Canny(image, 180,200)


# plotting

titles = ["original", "sobel","laplacian_all", "laplacian_abs", "canny"]
image_list = [image, sobel_combine, laplacian, laplacian_ab, canny]

plt.figure(figsize=(20,16))
for i in range(len(titles)):
    plt.subplot(2,3,i+1)
    plt.imshow(image_list[i], cmap='grey')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()    
```
